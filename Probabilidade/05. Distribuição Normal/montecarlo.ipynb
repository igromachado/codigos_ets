```python
import random as r
import math as m
import numpy as np
import matplotlib.pyplot as plt

n = 10000
atendimentos = []

for i in range(n):

    if i < 7000:
        numero = r.randint(5,20)
        atendimentos.append(numero)
    
    else:
        numero = r.randint(40,60)
        atendimentos.append(numero)

print("Média:", np.mean(atendimentos))
print("Desvio Padrão:", np.std(atendimentos))
```

O tempo médio está mais próximo de 20 minutos.

```python
grupo_3 = []

for i in range(n):
    grupo = r.sample(atendimentos, 3000)
    media_grupo = np.mean(grupo)
    grupo_3.append(media_grupo)

plt.hist(grupo_3, bins=30)
plt.title('Médias de grupos com 3 pacientes')
plt.xlabel('Tempo médio')
plt.ylabel('Frequência')
plt.show()
```

```python
import random as r
import numpy as np
import matplotlib.pyplot as plt

#QUESTÃO 1 - cálculo dos tempos de atendimento
def x():
    if r.random() < 0.15:
        return r.randint(40,60)
    else:
        return r.randint(5,20)
    
n = 10000
valores = []

for i in range(n):
    valores.append(x())

print("Média:", np.mean(valores))
print("Desvio Padrão:", np.std(valores))

#QUESTÃO 2 - cálculo das médias em grupos de atendimento
def y(n):
    soma = 0
    for i in range(n):
        soma += x()
    return soma/n

medias = []

for i in range(n):
    medias.append(y(100))

plt.hist(medias, bins=30)
plt.show()

#QUESTÃO 3 - cálculo de Z 
from scipy.stats import norm

media = 18
desvio = 4
x = 22

z = (x - media) / desvio
prob = 1 - norm.cdf(z)

print('A probabilidade de ser maior que 22 minutos é: ', prob)
```
