```python
import random as r
import numpy as np

simulacoes = 10000  
resultados = []

for simulacao in range(simulacoes):

    populacao = 100
    infectados = 1
    rodadas = 10

    for rodada in range(rodadas):

        for infectado in range(infectados):
            if infectados < populacao:
                if r.random() <= 0.30:
                    infectados += 1

    resultados.append(infectados)

sum = 0
for resultado in resultados:
    if resultado > 50:
        sum += 1

print("Média de infectados:", round(np.mean(resultados),2), "%")
print("Chance de metade da sala ser infectada:",round((sum/simulacoes)*100,2), "%")
```

R: Entre 13 a 14 pessoas.

b) Qual a chance de mais da metade da sala ser infectada?
R: Menos de 0.5% de chance.

c) Se repetirmos o experimento várias vezes, como se comporta a média de infectados?
R: Ela vai ficando cada vez mais exata.
