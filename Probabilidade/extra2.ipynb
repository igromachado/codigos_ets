```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('transacoes_fraude.csv')
```

```python
df = pd.read_csv('transacoes_fraude.csv')

soma = sum(df['fraude'])
proporção = round((soma / len(df))*100, 2)

print(proporção)
```

```python
df['suspeita'] = np.where((df['tentativas_login'] > 3) | (df['falha_login'] == 1), 1, 0)

soma = sum(df['suspeita'])
proporção = round((soma / len(df))*100, 2)

print(proporção)
```

```python
fraudes = df.groupby('fraude').agg(
    contagem = ('fraude', 'count')
).reset_index()

suspeitas = df.groupby('suspeita').agg(
    contagem = ('suspeita', 'count')
).reset_index()

fig, eixo = plt.subplots(nrows=1, ncols=2, figsize=(10,5))

eixo[0] = plt.bar(fraudes['fraude'], fraudes['contagem'])
eixo[1] = plt.bar(suspeitas['suspeita'], suspeitas['contagem'])
```
