9. Uma empresa afirma que 60% dos seus clientes estão satisfeitos com o serviço prestado. Um analista desconfia dessa afirmação e decide testá-la.
Ele coleta uma amostra de n = 80 clientes e observa que apenas 44 estão satisfeitos. Considere nível de significância de α = 5%.
a) Tome a decisão: rejeita ou não H0?
b) Gere 10.000 amostras simuladas de tamanho 80, assumindo que a hipótese nula é verdadeira:
• Cada cliente tem probabilidade p = 0,60 de estar satisfeito.
• Para cada simulação:
    o Conte quantos clientes estão satisfeitos
    o Calcule a proporção
• Estime a probabilidade de obter uma proporção menor ou igual a 44/80.
• Compare esse valor com α = 0,05.
O resultado da simulação confirma o teste teórico?

```python
import random
 
n = 10000
amostra = 80
 
satisfeito = []
proporcoes = []
 
for i in range(n):
    sat = 0
   
    for j in range(amostra):
        if random.random()<=0.6:
            sat+=1
           
    proporcao = sat/amostra
    satisfeito.append(sat)
    proporcoes.append(proporcao)
   
soma = 0
cont = 0
 
for i in range(n):
    soma+=satisfeito[i]
   
    if proporcoes[i]<=44/80:
        cont+=1
       
media = soma/n
prob = cont/n
prob_satisfeito = media/amostra
 
print('Média de satisfeitos:',round(media,2))
print('Probabilidade de satisfação:',round(prob_satisfeito,2))
print('Probabilidade de a proporção ser igual a 44/80:',round(prob,2))
```
