```python
import math as m
import matplotlib.pyplot as plt
import numpy as np
```

```python
questões = 20
lista = []

def pct(sucessos):

    fracassos = questões - sucessos

    chance_acerto = 1 / 5
    chance_fracasso = 1 - chance_acerto

    pct_acertos = chance_acerto ** sucessos
    pct_fracassos = chance_fracasso ** fracassos

    pct_total = pct_acertos * pct_fracassos

    fatorial = m.factorial(questões) / (m.factorial(sucessos) * m.factorial(fracassos))

    resultado = pct_total * fatorial

    lista.append(resultado)
    return resultado

sucessos = int(input('número de acertos:'))
pct(sucessos)
```

```python
lista = []

for i in range(9,21):
    pct(i)

print(sum(lista))
```

```python
print(f"Média total de acertos: {questões * 0.2}")
```

```python
lista = []

for i in range(1,21):
    pct(i)

lista = np.round(np.array(lista), 4)

x = np.arange(1,21)
y = lista

plt.figure(figsize=(15,5))
plt.plot(x, y, marker='o')
plt.xticks(x)
plt.xlabel("Número de acertos")
plt.ylabel('Probabilidade')
plt.title('Distribuição Binomial (n=20, p=0.2)')
plt.grid(linestyle='solid')

for i in range(len(x)):
    plt.text(x[i], lista[i], str(lista[i]))

plt.show()
```
