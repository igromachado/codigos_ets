```python
import random as r
import numpy as np

prob = 0.30
simulações = 10000
dias = 7
choveu_3 = 0
resultados = []

for i in range(simulações):
    choveu = 0
    for j in range(dias):
        if r.random() < prob:
            choveu += 1
    
    resultados.append(choveu)
    
    if choveu >= 3:
        choveu_3 += 1

print(f'Média de dias chovidos durante a semana: {round(np.mean(resultados),2)}')
print(f'Probabilidade de chover 3 dias na semana: {round((choveu_3 / simulações) * 100, 2)} %')
```

Chovem em média 2 dias na semana com a probabilidade dada. Ao analisar a probabilidade de chover 3 dias durante a semana, obtém-se o resultado aproximado de 34-35%, mostrando que as chances desse cenário acontecer seguem uma distruição normal.

```python
import math as m
import numpy as np
import random as r

prob = 0.5
lançada = 100
simulações = 10000
resultados = []
cont_65 = 0

for i in range(simulações):
    cara = 0

    for j in range(lançada):
        if r.random() <= prob:
            cara+=1

    resultados.append(cara)

    if cara >= 65:
        cont_65+=1

print(f'Média de caras: {round(np.mean(resultados), 2)}')
print(f'Quantidade de vezes em que saíram 65 caras ou mais: {cont_65}')
print(f'Probabilidade de saírem 65 caras: {round((cont_65 / simulações) * 100, 2)}%')
```

Não há indícios de que a moeda possa não ser justa, como a chance é de 50%, há casos em que seus valores serão menores ou maiores, podendo variar bastante. Ao analisar a probabilidade de saírem 65 caras após 100 jogadas - com 10000 simulações -, nota-se que as chances são bem pequenas (menor que 0), ou seja, o resultado apresentado foi uma exceção raríssima de acontecer.
