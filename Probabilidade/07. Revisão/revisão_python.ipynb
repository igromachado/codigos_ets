1) Crie um programa em Python que simule a retirada de 2 bolas de uma urna sem reposição. A urna possui:
• 4 bolas vermelhas
• 3 bolas azuis
• 2 bolas verdes
O programa deve:
a) Realizar 10.000 simulações da retirada de 2 bolas.
b) Contar quantas vezes:
• saem duas bolas da mesma cor;
• a segunda bola é vermelha sabendo que a primeira também foi vermelha.
c) Calcular:
• a probabilidade estimada de sair duas bolas da mesma cor;
• a probabilidade condicional da segunda bola ser vermelha dado que a primeira foi vermelha;
• a quantidade total de maneiras possíveis de retirar 2 bolas da urna utilizando combinação;

```python
import random as r
import numpy as np
import math as m

#a)
simulações = 10000
amostra = 2

cores_iguais = []
vermelhas_iguais = []
vermelhas_1 = []

for i in range(simulações):
    bolas = [0, 0, 0, 0, 1, 1, 1, 2, 2]
    c = 0
    v = 0
    v1 = 0
    retiradas = []

    for j in range(amostra):
        index = r.randint(0, len(bolas) - 1)
        retiradas.append(bolas[index])
        bolas.pop(index)

    if retiradas[0] == retiradas[1]:
        c+=1
    if retiradas[0] == 0:
        v1+=1
        if retiradas[1] == 0:
            v+=1

    cores_iguais.append(c)
    vermelhas_iguais.append(v)
    vermelhas_1.append(v1)

#b)
print(f"Contagem de cores iguais: {np.sum(cores_iguais)}")
print(f"Contagem de vermelhas iguais: {np.sum(vermelhas_iguais)}")

print()

#c)
print(f"Probabilidade estimada de sair duas bolas da mesma cor: {round(np.mean(cores_iguais) * 100, 2)}%")
print(f"Probabilidade condicional da segunda bola ser vermelha dado que a primeira foi vermelha: {round(np.sum(vermelhas_iguais) / np.sum(vermelhas_1) * 100, 2)}%")
print(f"A quantidade total de maneiras possíveis de retirar 2 bolas da urna utilizando combinação: {m.comb(9,2)}")


```

2) Uma escola afirma que 75% dos seus alunos foram aprovados em uma prova de matemática. Um coordenador pedagógico desconfia dessa afirmação e decide testá-la. Ele seleciona uma amostra aleatória de 120 alunos e observa que apenas 81 alunos foram aprovados. Considere nível de significância de α = 5%.
a) Realize o teste de hipótese para verificar se a proporção de aprovados é realmente igual a 75%.
b) Desenvolva um programa em Python que gere 10.000 amostras simuladas de tamanho 120, assumindo que a hipótese nula seja verdadeira. Considere que cada aluno possui probabilidade de 75% de ser aprovado.
Para cada simulação:
• Gere os 120 resultados (aprovado ou reprovado);
• Conte quantos alunos foram aprovados;
• Calcule a proporção de aprovados.
Depois:
• Estime a probabilidade de obter uma proporção menor ou igual a 81/120 conforme a amostra selecionada pelo coordenador.
• Compare essa probabilidade com 𝛼=0,05 para aceitar ou rejeitar o teste teórico.

```python
import random as r
import numpy as np

#b)
simulacao = 10000
amostra = 120
aprovados = []
proporções = []

for i in range(simulacao):
    ap = 0 

    for j in range(amostra):
        if r.random() <= 0.75:
            ap +=1

    proporção = ap/amostra
    aprovados.append(ap)
    proporções.append(proporção)

print(f"Quantidade de alunos aprovados: {np.sum(aprovados)}")
print(f"Proporção de alunos aprovados: {round(np.mean(aprovados) / amostra,2)}")

#proporção de 81/120
cont = 0
for i in range(simulacao):
    if proporções[i] <= 81/120:
        cont+=1

prob_81_120 = cont / simulacao

print(f"Estime a probabilidade de obter uma proporção menor ou igual a 81/120 conforme a amostra selecionada pelo coordenador: {round(prob_81_120, 2)}")
```
