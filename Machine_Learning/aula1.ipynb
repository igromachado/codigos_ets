```python
import pandas as pd
import math as m


df = pd.read_csv('plano_saude.csv')
n = len(df) - 1

def funcaoCv(df):
    somatorio = 0

    for i in range(len(df)):

        x = df['idade'][i]
        y = df['custo'][i]

        mx = df['idade'].mean()
        my = df['custo'].mean()

        somatorio += (x - mx) * (y - my)
    
    cv = somatorio / n

    return cv

def funcaoCr(cv):
    x = m.std(df['idade'])
    y = m.std(df['custo'])

    cr = cv / (x * y)

    return cr


v = int(input('Digite o valor que quer prever:'))

cv = funcaoCv()
```
