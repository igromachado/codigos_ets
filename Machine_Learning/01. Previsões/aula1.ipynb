MACHINE LEARNING

COVARIÂNCIA -> C(x,y) = somatório (xi - médiaX) * (yi - médiay) / n - 1
CORRELAÇÃO -> cr = cv / (std(x) * std(y))
INCLINAÇÃO -> m = cr * (std(y) / std(x))
INTERSEÇÃO -> b = médiaY - m * médiaX
PREVISÃO -> p = b * m * v
V = valor que quero preencher

correlação > inclinação > interseção > previsão

```python
import pandas as pd
import math as m
import numpy as np


df = pd.read_csv('plano_saude.csv')
n = len(df) - 1

def funcaoCv(df):
    somatorio = 0

    for i in range(len(df)):

        x = df['idade'][i]
        y = df['custo'][i]

        mx = df['idade'].mean()
        my = df['custo'].mean()

        somatorio += (x - mx) * (y - my)
    
    cv = somatorio / n

    return cv

def funcaoCr(df):
    x = np.std(df['idade'], ddof=1)
    y = np.std(df['custo'], ddof=1)

    cr = funcaoCv(df) / (x * y)

    return cr

def funcaoM(df):
    x = np.std(df['idade'], ddof=1)
    y = np.std(df['custo'], ddof=1)

    m = funcaoCr(df) * (y / x)

    return m

def funcaoB(df):
    x = df['idade'].mean()
    y = df['custo'].mean()

    b = y - (funcaoM(df) * x)

    return b

def funcaoP(df, v):
    p = funcaoB(df) + (funcaoM(df) * v)

    return float(p)

v = 54

funcaoP(df, v)
```

```python
from sklearn.linear_model import LinearRegression
import pandas as pd
import numpy as np

df = pd.read_csv('plano_saude.csv')

x = df['idade']
y = df['custo']

x = x.values.reshape(-1,1)

regressor_plano_saude = LinearRegression()
regressor_plano_saude.fit(x,y)
```
