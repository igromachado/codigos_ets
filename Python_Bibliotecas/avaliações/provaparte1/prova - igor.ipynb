```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
temperaturas = np.random.randint(18,41,(7,4))
print(temperaturas)
```

```python
print(f"Formato: {temperaturas.shape}")
print(f"Número de dimensões: {temperaturas.ndim}")
print(f"Quantidade total de elementos: {temperaturas.size}")
```

```python
novo = np.array(temperaturas)

temperaturas_novo = novo.reshape(4,7)
print(temperaturas_novo)
```

```python
temperaturas_corrigidas = temperaturas_novo + 1.5

print(temperaturas_corrigidas)
```

```python
print(f"Média de temperatura por máquina: {temperaturas_corrigidas.mean(axis=1)}")
print(f"Média de temperatura por dia: {temperaturas_corrigidas.mean(axis=0)}")
```

```python
print(f"Temperaturas da máquina 2:\n {temperaturas_corrigidas[1:2,:]} \n")
print(f"Temperaturas registradas no dia 5: \n {temperaturas_corrigidas[:,4:5]} \n")
print(f"Temperaturas das máquinas 1 a 3 no dia 4: \n {temperaturas_corrigidas[0:3,3:4]} \n")
```

```python
acima_35 = temperaturas_corrigidas[temperaturas_corrigidas>35]

print(f"Temperaturas acima dos 35 graus: {acima_35}")
```

```python
menor_22 = np.array(temperaturas_corrigidas[temperaturas_corrigidas<22])

print(f"Temperaturas menores que 22 graus: {menor_22}")
```

```python
print(f"Média geral: {round(temperaturas_corrigidas.mean(),2)}")
print(f"Valor máximo: {temperaturas_corrigidas.max()}")
print(f"Valor mínimo: {temperaturas_corrigidas.min()}")
print(f"Desvio padrão: {round(temperaturas_corrigidas.std(),2)}")
```

```python
vendedores = [1,2,1,3,2,1,3,2,3,1]
regiões = ["Norte","Sul","Norte","Sul","Leste","Norte","Leste","Sul","Sul","Leste"]
vendas = [120,200,150,300,180,90,250,210,310,160]

vendedores = np.array(vendedores)
regiões = np.array(regiões)
vendas = np.array(vendas)
```

```python
por_vendedor = np.array([vendedores, vendas])

um = np.where(por_vendedor[:1] == 1, por_vendedor, False)
dois = np.where(por_vendedor[:1] == 2, por_vendedor, False)
tres = np.where(por_vendedor[:1] == 3, por_vendedor, False)

print(f"Total de vendas por vendedor:")
print(f"Vendedor 1: {um[1:2].sum()}")
print(f"Vendedor 2: {dois[1:2].sum()}")
print(f"Vendedor 3: {tres[1:2].sum()}")
```

```python
por_região = np.array([regiões, vendas])

leste = np.where(por_região[:1] == 'Leste', por_região, 0)
norte = np.where(por_região[:1] == 'Norte', por_região, 0)
sul = np.where(por_região[:1] == 'Sul', por_região, 0)

print(f"Total de vendas por região:")
print(f"Região Leste: {leste[1:2].astype(int).sum()}")
print(f"Região Norte: {norte[1:2].astype(int).sum()}")
print(f"Região Sul: {sul[1:2].astype(int).sum()}")
```

```python
print(f"Média geral de vendas: {vendas.mean()}")
print(f"Vendas acima da média: \n {vendas[vendas>vendas.mean()]}")
```

```python
python = np.array([101,102,103,104,105,106,107])
estatistica = np.array([103,104,108,109,110,105])
visualizacao = np.array([101,103,105,111,112])
```

```python
py_est = np.union1d(python,estatistica)
print(f"Python ou Estatística: {py_est}")
```

```python
tres_cursos = np.intersect1d(python,estatistica)
tres_cursos = np.intersect1d(tres_cursos,visualizacao)
print(f"Nos três cursos: {tres_cursos}")
```

```python
est_vis = np.union1d(estatistica, visualizacao)
est_vis = np.setdiff1d(est_vis,python)
print(f"Estatística ou Visualização, mas não Python: {est_vis}")
```

```python
df = pd.read_csv('consumo_de_energia_cidades.csv')
df.head()
```

```python
CaP = df[df['Cidade'] == 'Campinas']

plt.plot(CaP.Mes, CaP.Consumo_Residencial, marker = 'o', label='Consumo Residencial')
plt.plot(CaP.Mes, CaP.Consumo_Comercial, marker = 'o', label= 'Consumo Comercial')
plt.title('Consumo de Energia em Campinas por Mês')
plt.xlabel('Mês')
plt.ylabel('Consumo de Energia (MWh)')
plt.legend()
plt.show()
```

```python
anual = df.groupby('Cidade')['Consumo_Residencial'].sum().reset_index()

#só pra deixar colorido bonitinho
campinas = anual[anual['Cidade'] == 'Campinas']
Ribeirao_Preto = anual[anual['Cidade'] == 'Ribeirao_Preto']
Sorocaba = anual[anual['Cidade'] == 'Sorocaba']

fig, eixo = plt.subplots(nrows = 1, ncols = 2, figsize = (15,5))

eixo[0].pie(anual.Consumo_Residencial, labels=anual.Cidade, autopct= "%1.1f%%")
eixo[0].set_title('Participação Percentual no Consumo Residencial Total')

eixo[1].bar(campinas.Cidade, campinas.Consumo_Residencial)
eixo[1].bar(Ribeirao_Preto.Cidade, Ribeirao_Preto.Consumo_Residencial)
eixo[1].bar(Sorocaba.Cidade, Sorocaba.Consumo_Residencial)
eixo[1].set_title('Consumo Residencial Total Anual por Cidade')
eixo[1].set_xlabel('Cidade')
eixo[1].set_ylabel('Consumo Total Residencial')

plt.show()
```

```python
plt.hist(df.Temperatura_Media, edgecolor="black")
plt.title('Distribuição da Temperatura Média nas Cidades')
plt.xlabel('Temperatura Média (°C)')
plt.ylabel('Frequência')
plt.show()
```

```python
fig, eixo = plt.subplots(nrows = 2, ncols = 1, figsize = (10,10))

x = np.arange(1,7,1)

eixo[0].plot(x ** x, color='green')
eixo[0].set_ylabel('x!')
eixo[0].set_title('Função f(x) = x!')
eixo[0].grid(linestyle='solid')

eixo[1].plot(x*x, color='red')
eixo[1].set_ylabel('x²')
eixo[1].set_title('Função g(x) = x²')
eixo[1].grid(linestyle='solid')

eixo[0].tick_params(axis='x')
eixo[1].tick_params(axis='x')

plt.show()
```

```python
df = pd.read_csv('desempenho_estudantes.csv')
df.head()
```

```python
sns.relplot(data=df, x='Horas_Estudo', y='Nota_Estatistica', hue='Curso')
plt.xlabel('Horas de Estudo')
plt.title('Relação entre Horas de Estudo e Nota de Estatística')
plt.show()
```

```python
media = df.groupby('Participacao')['Nota_Estatistica'].mean().reset_index().sort_values('Nota_Estatistica', ascending=False)

sns.barplot(data=media, x='Participacao', y='Nota_Estatistica')
plt.ylabel('Média da Nota de Estatística')
plt.xlabel('Participacao')
plt.title('Média da Nota de Estatística por Participação')
plt.show()
```

```python
sns.histplot(data=df, x='Nota_Matematica', kde=True)
plt.xlabel('Nota de Matemática')
plt.ylabel('Frequência')
plt.title('Distribuição da Nota de Matemática')
plt.show()
```
