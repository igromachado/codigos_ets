```python
df = pd.read_csv('ecommerce.csv')
df.head()
```

```python
sns.histplot(data=df, x='Valor_Compra', kde=True, edgecolor='white')
plt.title('Distribuição do Valor de Compra')
plt.grid(linestyle='solid')
plt.ylabel('Frequência')
plt.show()
```

A maioria das compras são de médio valor.

```python
sns.boxplot(data=df, x='Categoria_Produto', y='Valor_Compra')
plt.grid(linestyle='solid')
plt.xlabel('Categoria')
plt.ylabel('Valor da compra')
plt.show()
```

Todas as categorias possuem variações semelhantes, as com um pouco mais de variação seriam as categorias 'Casa' e 'Roupas', por conta do tamanho do boxplot.

```python
sns.scatterplot(data=df, x='Quantidade_Itens', y='Valor_Compra', hue='Categoria_Produto')
plt.xlabel('Quantidade de Itens')
plt.ylabel('Valor de Compra')
plt.title('Relação entre Quantidade de Itens e Valor de Compra')
plt.show()
```

Quanto mais na faixa média de valores, maior a quantidade de itens comprados.

```python
sns.countplot(data=df, x='Metodo_Pagamento')
plt.title('Métodos de Pagamento')
plt.ylabel('Quantidade')
plt.xlabel('Método')
plt.show()
```

O pagamento por Pix é o mais utilizado nas compras.

```python
with open('pedidos.json', 'r') as f:
    velha_estrutura = json.load(f)
    
teste = []
 
for i in velha_estrutura:
    
    id_pedido = i['id_pedido']
    valor_total = 0
        
    for chave, valor in i['cliente'].items():
        if chave == 'nome':
            nome_cliente = valor
    
    for k in i['itens']:
        quantidade = k['preco'] * k['quantidade']
        valor_total += quantidade
            
    nova_estrutura = {'id_pedido': i['id_pedido'], 'nome_cliente': nome_cliente, 'valor_total': valor_total}
    teste.append(nova_estrutura)

with open('nova_estrutura.json', 'w') as f:
    json.dump(teste, f, indent=4)
    
print('Nova estrutura salva com sucesso!')
```

```python
dir = os.path.abspath('')
pasta = 'PDFs'

if not os.path.isdir(pasta):
    os.mkdir(pasta)
else:
    print('Pasta já existe no diretório!')
    
arquivos = os.listdir()

cont = 0

for i in arquivos:
    if i[-4:] == '.pdf':
        origem = os.path.join(dir,i)
        destino = os.path.join(dir,pasta,i)
        os.rename(origem,destino)      
        cont+=1

if cont > 0:
    print(f"{cont} arquivos movidos para a pasta 'PDFs'.")
else:
    print("Nenhum arquivo '.pdf' foi encontrado!")  
```

```python
dir = os.path.abspath('')
pasta = 'logs'
arq = 'acessos.json'

#criar a pasta
if not os.path.isdir(pasta):
    os.mkdir(pasta)

#pegar o caminho do arquivo
dirpath = os.path.join(dir,pasta,arq)

#pegar o tempo atual
tempo_atual = time.time()
tempo = time.localtime()
data_hora = (f"{tempo.tm_mday}/{tempo.tm_mon}/{tempo.tm_year} {tempo.tm_hour}:{tempo.tm_min}:{tempo.tm_sec}")


#cria o arquivo se ele não existir
if not os.path.isfile(dirpath):
    with open (dirpath, 'w') as f:
        dados = [{'timestamp':tempo_atual, 'data_hora': data_hora}] #sempre se inicia como uma lista, para que seja possível dar append em outros acessos
        json.dump(dados, f, indent=4)
        
#adicionar itens no json caso o arquivo já exista
else:
    with open(dirpath, 'r') as f: 
        dados = json.load(f) #lê o conteúdo presente no json
        
    jason = {'timestamp':tempo_atual, 'data_hora': data_hora}
    dados.append(jason) #dá um append com o acesso atual no json antigo

    with open (dirpath, 'w') as f:
        json.dump(dados, f, indent=4)

print(f'Acesso atual: {data_hora}')
print(f"Total de acessos registrados: {len(dados)}")
```

```python
import random as r

cap = cv2.VideoCapture('babydance.mp4')
fourcc = cv2.VideoWriter_fourcc(*'XVID')
output = cv2.VideoWriter('questão7.mp4', fourcc, 20.0, (500,500))
font = cv2.FONT_HERSHEY_SIMPLEX

while True:
    ret, frame = cap.read()
    if ret == True:
        frame = cv2.flip(frame,0)
        frame = cv2.resize(frame, (500,500))
        cv2.putText(frame, 'dance baby, dance!', (r.randint(0,250),r.randint(0,500)), font, 1.5, (191,0,191), 4, cv2.LINE_AA)
        cv2.imshow('teste',frame)
        output.write(frame)
        cv2.waitKey(1)
    else:
        break
    
cap.release()
output.release()
cv2.destroyAllWindows()
```

```python
img = cv2.imread('underwater.jpg')

print('Escolha o efeito de cor: \n1 - BGR para Luv\n2 - BGR para HSV\n3 - BGR para RGB')

cor = int(input('Escolha o efeito de cor (1/2/3)'))

match cor:
    case 1:
        img = cv2.cvtColor(img, cv2.COLOR_BGR2LUV)
    case 2:
        img = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    case 3:
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    case _:
        print('Opção inválida')

print(f"Opção: {cor}\n")

print('Escolha o tipo de espelhamento: \n1 - Manter original \n2 - Inverter horizontalmente\n3 - Inverter verticalmente')

espelhamento = int(input('Escolha o tipo de espelhamento (1/2/3)'))

match espelhamento:
    case 1:
        img = img
    case 2:
        img = cv2.flip(img, 1)
    case 3:
        img = cv2.flip(img, 0)
        
print(f"Opção: {espelhamento}\n")

cv2.imwrite('nova_imagem.jpg', img)

print("Imagem processada e salva como 'nova_imagem.jpg'")
```
